6. Maior valor e sua posicao

```python
maior = 0
posicao = 1

for i in range(1, 9):
    num = int(input(f"Digite o {i}º número: "))
    if i == 1:
        maior = num
        posicao = 1
    else:
        if num > maior:
            maior = num
            posicao = i

print(f"Maior valor: {maior}")
print(f"Posição da primeira vez que apareceu: {posicao}")
```

7.Segundo maior valor

```python

maior = 0
segundo_maior = 0

for i in range(8):
    num = int(input(f"Digite o {i+1}º número: "))
    if i == 0:
        maior = num
    elif i == 1:
        if num > maior:
            segundo_maior = maior
            maior = num
        else:
            segundo_maior = num
    else:
        if num > maior:
            segundo_maior = maior
            maior = num
        elif num > segundo_maior:
            segundo_maior = num

print(f"Maior: {maior}")
print(f"Segundo maior: {segundo_maior}")
```

8. Detectando repeticao consecutiva

```python


tem_igual = False
anterior = int(input("Digite o 1º número: "))

for i in range(7):
    atual = int(input(f"Digite o {i+2}º número: "))
    if atual == anterior:
        tem_igual = True
    anterior = atual

if tem_igual:
    print("Existem valores consecutivos iguais!")
else:
    print("Não existem valores consecutivos iguais.")
```

9. Controle de saldo

```python

saldo = 500.00

for i in range(6):
    print(f"\nOperação {i+1}")
    tipo = int(input("1 - Depósito | 2 - Saque: "))
    valor = float(input("Valor: R$ "))

    if tipo == 1:
        saldo += valor
    elif tipo == 2:
        if valor <= saldo:
            saldo -= valor
        else:
            print("Saldo insuficiente!")

print(f"\nSaldo final: R$ {saldo:.2f}")
```

10. Caixa eletrônico simplificado

```python

valor = int(input("Digite o valor em reais: R$ "))

cedulas = [100, 50, 20, 10, 5, 2, 1]

print("\nQuantidade de cédulas:")
for c in cedulas:
    qtd = valor // c
    if qtd > 0:
        print(f"R$ {c}: {qtd} cédula(s)")
    valor = valor % c
```

11. Número primo

```python

n = int(input("Digite um número maior que 1: "))
eh_primo = True

if n > 1:
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            eh_primo = False
            break

if eh_primo:
    print(f"{n} é um número primo!")
else:
    print(f"{n} não é um número primo.")
```

12. Divisores de um numero

```python

n = int(input("Digite um número inteiro positivo: "))
contador = 0

print(f"Divisores de {n}:")
for i in range(1, n + 1):
    if n % i == 0:
        print(i)
        contador += 1

print(f"Total de {contador} divisores.")
```

13. Numeros primos em um intervalo

```python

A = int(input("Digite o valor de A: "))
B = int(input("Digite o valor de B: "))
contador = 0

print(f"Números primos entre {A} e {B}:")

for num in range(A, B + 1):
    if num > 1:
        primo = True
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                primo = False
                break
        if primo:
            print(num)
            contador += 1

print(f"Total de {contador} números primos encontrados.")
```

14. Jogo com limite de tentativas

```python

SECRETO = 37
tentativas = 0
acertou = False

while tentativas < 5 and not acertou:
    palpite = int(input("Digite seu palpite: "))
    tentativas += 1

    if palpite == SECRETO:
        print("Parabéns! Você acertou! 🎉")
        acertou = True
    elif palpite > SECRETO:
        print("O número secreto é MENOR.")
    else:
        print("O número secreto é MAIOR.")

if not acertou:
    print(f"Fim de jogo! O número era {SECRETO}.")
```

15. Desafio final- caixa de atendimento

```python

saldo = 200.00
qtd_entradas = 0
qtd_retiradas = 0
qtd_recusadas = 0
total_entrada = 0.00
total_saida = 0.00

while True:
    print("\n--- MENU ---")
    print("1 - Entrada de dinheiro")
    print("2 - Retirada de dinheiro")
    print("0 - Encerrar")
    opcao = int(input("Escolha uma operação: "))

    if opcao == 0:
        break
    elif opcao == 1:
        valor = float(input("Valor da entrada: R$ "))
        saldo += valor
        total_entrada += valor
        qtd_entradas += 1
    elif opcao == 2:
        valor = float(input("Valor da retirada: R$ "))
        if valor <= saldo:
            saldo -= valor
            total_saida += valor
            qtd_retiradas += 1
        else:
            print("Saldo insuficiente — retirada recusada!")
            qtd_recusadas += 1
    else:
        print("Opção inválida!")

print("\n" + "="*35)
print(f"Saldo final:      R$ {saldo:.2f}")
print(f"Entradas:         {qtd_entradas} | Total: R$ {total_entrada:.2f}")
print(f"Retiradas:        {qtd_retiradas} | Total: R$ {total_saida:.2f}")
print(f"Retiradas recusadas: {qtd_recusadas}")
print("="*35)
```
