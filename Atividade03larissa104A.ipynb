1. Guardando e analisando notas

```python
notas = []
for i in range(5):
    nota = float(input("Digite a nota: "))
    notas.append(nota)

soma = 0
for nota in notas:
    soma += nota

media = soma / len(notas)

acima_ou_igual = 0
for nota in notas:
    if nota >= media:
        acima_ou_igual += 1

print("Notas:", notas)
print("Média:", media)
print("Notas >= média:", acima_ou_igual)
```

2. Coordenada representada por tupla

```python
x = int(input("x: "))
y = int(input("y: "))

ponto = (x, y)

if ponto[0] == 0 and ponto[1] == 0:
    local = "Origem"
elif ponto[0] == 0:
    local = "Eixo Y"
elif ponto[1] == 0:
    local = "Eixo X"
elif ponto[0] > 0 and ponto[1] > 0:
    local = "1º quadrante"
elif ponto[0] < 0 and ponto[1] > 0:
    local = "2º quadrante"
elif ponto[0] < 0 and ponto[1] < 0:
    local = "3º quadrante"
else:
    local = "4º quadrante"

print("Ponto:", ponto)
print("Local:", local)
```

4. Lista de números: pares e ímpares

```python

numeros = []
pares = []
impares = []
for i in range(10):
    n = int(input(f"Digite o {i+1}º número: "))
    numeros.append(n)
for num in numeros:
    if num % 2 == 0:
        pares.append(num)
    else:
        impares.append(num)
print("\n Lista completa:", numeros)
print("Números pares:", pares)
print("Números ímpares:", impares)
```

5. Maior e menor da lista

```python
numeros = []
for i in range(8):
    num = int(input(f"Digite o {i+1}º número: "))
    numeros.append(num)

maior = numeros[0]
menor = numeros[0]
pos_maior = 0
pos_menor = 0

for i in range(1, 8):
    if numeros[i] > maior:
        maior = numeros[i]
        pos_maior = i
    if numeros[i] < menor:
        menor = numeros[i]
        pos_menor = i

print("\nLista:", numeros)
print(f"Maior valor: {maior} na posição {pos_maior}")
print(f"Menor valor: {menor} na posição {pos_menor}")
```

7. Dias da semana em uma tupla

```python
t = []
for i in range(6):
    t.append(int(input("Numero: ")))
t = tuple(t)

cont5 = 0
pos5 = -1
pares = 0

for i in range(6):
    if t[i] == 5:
        cont5 = cont5 + 1
        if pos5 == -1:
            pos5 = i
    if t[i] % 2 == 0:
        pares = pares + 1

print("5 apareceu", cont5)
print("Pos", pos5)
print("Pares", pares)
```

12. Estoque e atualização

```python
est = {"arroz":10, "feijao":5, "acucar":8, "cafe":3}
nome = input("Produto: ")
qtd = int(input("Qtd: "))
if nome in est:
    if qtd <= est[nome]:
        est[nome] = est[nome] - qtd
        print("Ok")
    else:
        print("Sem estoque")
else:
    print("Nao existe")
print(est)
```

15. Mini sistema de estoque

```python
estoque = []
for i in range(5):
    codigo = int(input("Codigo: "))
    nome = input("Nome: ")
    qtd = int(input("Qtd: "))
    preco = float(input("Preco: "))

    item = {"codigo": codigo, "nome": nome, "quantidade": qtd, "preco": preco}
    estoque.append(item)
for item in estoque:
    print(item)
total = 0
for item in estoque:
    total = total + item["quantidade"] * item["preco"]
print("Total:", total)
maior = estoque[0]
for item in estoque:
    if item["quantidade"] * item["preco"] > maior["quantidade"] * maior["preco"]:
        maior = item
print("Maior:", maior["nome"])
resumo = (5, total)
print(resumo)
```
