#5.situacao do aluno




```python
nota1 = float(input("Nota 1: "))
nota2 = float(input("Nota 2: "))
media = (nota1 + nota2) / 2

print(f"Média: {media:.1f}")
if media >= 7:
    print("Aprovado")
elif media >= 5:
    print("Recuperação")
else:
    print("Reprovado")
```

#8. Somando de 1 até N

```python
n = int(input("Digite N: "))
soma = 0

for i in range(1, n+1):
    soma = soma + i

print("Soma =", soma)
```

#14. Número secreto

```python
secreto = 7
tentativas = 0
palpite = 0

while palpite != secreto:
    palpite = int(input("Tente adivinhar: "))
    tentativas = tentativas + 1
    if palpite < secreto:
        print("É maior")
    elif palpite > secreto:
        print("É menor")

print("Acertou em", tentativas, "tentativas!")
```

#15. Analisando uma sequência de números

```python
quantidade = 0
positivos = 0
negativos = 0
soma = 0

while True:
    num = int(input("Número: "))
    if num == 0:
        break
    quantidade = quantidade + 1
    soma = soma + num
    if num > 0:
        positivos = positivos + 1
    else:
        negativos = negativos + 1

print("Quantidade =", quantidade)
print("Positivos =", positivos)
print("Negativos =", negativos)
print("Soma =", soma)
```
